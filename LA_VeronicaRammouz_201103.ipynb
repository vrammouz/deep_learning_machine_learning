## Task 1:- Find the box[20 points]

#### Suppose that we have three colored boxes r (red), b (blue), and g (green).
#### - Box r contains 3 apples, 4 oranges, and 3 limes,
#### - Box b contains 1 apple, 1 orange, and 0 limes, and
#### - Box g contains 3 apples, 3 oranges, and 4 limes.
#### soln1: If a box is chosen at random with probabilities p(r) = 0.2, p(b) = 0.2, p(g) = 0.6, and a piece of fruit is removed from the box (with equal probability of selecting any of the items in the box), then what is the probability of selecting an apple?
#### soln2: If we observe that the selected fruit is, in fact an orange, what is the probability that it came from the green box?

**Give your mathematical writing here**

```python
#Box probabilities
p_r = 0.2
p_b = 0.2
p_g = 0.6

#Fruit probabilities
p_apples_given_r = 3/10
p_apples_given_b = 1/2
p_apples_given_g = 3/10

p_oranges_given_r = 4/10
p_oranges_given_b = 1/2
p_oranges_given_g = 3/10

p_g_given_orange = 1/3

#Computed fruit probabilities
p_apple = (p_apples_given_r * p_r) + (p_apples_given_b * p_b) + (p_apples_given_g * p_g)

p_orange = (p_oranges_given_r * p_r) + (p_oranges_given_b * p_b) + (p_oranges_given_g * p_g)

#Solutions
soln1 = p_apple
soln2 = (p_g_given_orange * p_g) / p_orange

print(soln1)
print(soln2)
```

## Task 2:- Let's get into magic[20 points]

#### Assume we have 3 cards mixed up in the hat which are identical in form but differs in the color,
#### - the first card is colored green on both sides,
#### - the second card is colored green on one side and blue on the other side
#### - the third card is colored blue on both sides.

#### Now one card is randomly selected, which has green color on one side. What is the probability that the other side is blue?

##### **Hint**: Monty Hall problem


**Give your mathematical writing here**

```python
#Probabilities
p_blue = 1/3

p_green = 1/3

p_not_blue = p_green

p_green_given_blue = 1/2
p_green_given_not_blue = 1

#Solution
soln3 = (p_green_given_blue * p_blue) / ((p_green_given_blue * p_blue) + (p_green_given_not_blue * p_not_blue))

print(soln3)
```

## Task 4:- Gaussian Distribution[40 points]

#### Consider a Gaussian distribution $p_x(x)$ over x with mean μ = 6 and standard deviation $\sigma$ = 1. Plot this distribution.

```python
#Mu and sigma
mu = 6
sig = 1

#Gaussian function
def gaussian(x, mu, sig):
    return ((1 / (np.sqrt(2 * np.pi) * sig**2)) * np.exp(-0.5 * ((x - mu)/ sig)**2))

#Define x values and derive y
x = np.linspace(0, 10, 50000)
y = gaussian(x, mu, sig)

#Plot figure with custom attributes
plt.figure(figsize=(15, 8))
plt.plot(x, y, color='green')
plt.grid()
plt.xlim(0,10)
plt.ylim(0,1)
plt.title('Gaussian Distribution')

plt.show()
```

#### Draw a sample of N = 50000 points from this distribution and plot a histogram of their values, which as expected agrees with the distribution $p_x(x)$.

```python
#Values
N = 50000
num_bins = 50

x_hist = mu + sig * np.random.randn(N)

fig, ax = plt.subplots(figsize=((15,8)))

#Histogram data
n, bins, patches = ax.hist(x_hist, num_bins, density=1, color='indigo', edgecolor='black')

#Redefine Gaussian curve
y = ((1 / (np.sqrt(2 * np.pi) * sig**2)) * np.exp(-0.5 * ((bins - mu)/ sig)**2))

#Plot data
ax.plot(bins, y, color='green')
ax.set_title(r'Gaussian Distribution with $\mu=6$ and $\sigma=1$ and Histogram for N=50000 Points')
plt.grid()
plt.xlim(0,10)
plt.ylim(0,1)
plt.show()
```

#### Now consider a non-linear change of variables from x to y given by  
#### $x=g(y)=ln(y)-ln(1-y)+5$  
#### The inverse of this function is given by  
#### $y=g^{-1}(x)=\frac{1}{1+exp(-x+5)}$  
#### which is a logistic sigmoid function. Plot this function along with the previous plot

```python
#Define sigmoid function
def sigmoid(x):
    return 1/(1 + np.exp(-x + 5))

#Define x and y
x_sigmoid = np.linspace(0, 10, 100)
y_sigmoid = sigmoid(x_sigmoid)

#Plot Gaussian curve again
fig, ax = plt.subplots(figsize=((15,8)))

n, bins, patches = ax.hist(x_hist, num_bins, density=1, color='indigo', edgecolor='black')
ax.plot(bins, y, color='green')

#Plot Sigmoid curve
plt.plot(x_sigmoid, y_sigmoid, color='red')

#Plot attributes
plt.title('Gaussian and Sigmoid Curves')
plt.grid()

plt.show()
```

#### Plot the curve $p_x(g(y))$ that is a transformation of $p_x(x)$ as a function of x, $p_x(g(y))$ along with the previous plots, this should look exactly as the above one(reference provided above).

```python
#Define logit function g(y)
def logit(y):
    return (np.log(y) - np.log(1-y) + 5)

y_logit = np.linspace(0, 1, 100)
y_logit = y_logit[(y_logit > 0) & (y_logit < 1)]

# y_l = sigmoid(y_logit)

g = logit(y_logit)

#Compute transformation function p(g(y)), given p(x) is the gaussian distribution function
transformation = gaussian(g, mu, 0.4)

fig, ax = plt.subplots(figsize=((15,8)))

#Plot Gaussian curve and histogram again
n, bins, patches = ax.hist(x_hist, num_bins, density=1, color='indigo', edgecolor='black', label='Histogram')
ax.plot(bins, y, color='green', label="p(x)")

#Plot Sigmoid curve
plt.plot(x_sigmoid, y_sigmoid, color='red', label="y(x)")

#Plot Transformation curve
plt.plot(transformation, y_logit, '--', color='turquoise', label="p(g(y))")

#Plot attributes
plt.title('Gaussian Distribution, Sigmoid Function and Transformation')
plt.grid()
plt.legend(loc="upper left")
plt.ylim(0, 1)
plt.show()
```
